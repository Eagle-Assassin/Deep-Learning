```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from  tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense,Bidirectional


#load the IMDB Dataset
(X_train,y_train),(x_test,y_test)=imdb.load_data(num_words=1000)

#Pad sequence  to have the same length
X_train=pad_sequences(X_train,maxlen=100)
X_test=pad_sequences(x_test,maxlen=100)


#Define  the RNN Model

model= Sequential([Embedding(10000,32,input_shape=(100,)), #Embedding layers  to convert  words to vectors
                   SimpleRNN(5,return_sequences=True),    #RNN layers  with 5 units
                   SimpleRNN(5),                          #Another RNN  layers  witj 5 units
                   Dense(1,activation='sigmoid')          #Outputs  layer for binary  classification
                ])       
model.summary() 

model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])

history = model.fit(X_train,y_train,epochs=5,batch_size=32,validation_split=0.2)

```

```python
#Define  the Bidirectional Model

model= Sequential([Embedding(10000,32,input_shape=(100,)), #Embedding layers  to convert  words to vectors
                   Bidirectional(SimpleRNN(5)),                          #Another RNN  layers  witj 5 units
                   Dense(1,activation='sigmoid')          #Outputs  layer for binary  classification
                ])       
model.summary() 
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])

history = model.fit(X_train,y_train,epochs=5,batch_size=32,validation_split=0.2)
```
