```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform, expon,norm,laplace,binom,geom,multivariate_normal as mvn
```

```python
xvals= np.linspace(-2,5,1000)
yvals=uniform.cdf(xvals,1,2)

plt.plot(xvals,yvals)
plt.title('Uniform (1,3)')
plt.ylabel('F(x)')
plt.xlabel('x')
```

```python
xvals= np.linspace(-2,5,1000)
yvals=expon.cdf(xvals,1,2)

plt.plot(xvals,yvals)
plt.title('expon (1,3)')
plt.ylabel('F(x)')
plt.xlabel('x')
```

```python
xvals= np.linspace(-20,50,1000)
yvals=norm.cdf(xvals,10,5)

plt.plot(xvals,yvals)
plt.title('N (10,5^2)')
plt.ylabel('F(x)')
plt.xlabel('x')
```

```python
xvals= np.linspace(-20,50,1000)
yvals=laplace.cdf(xvals,10,5)

plt.plot(xvals,yvals)
plt.title('Laplace 10,5)')
plt.ylabel('F(x)')
plt.xlabel('x')
```

```python
xvals= np.arange(100)
yvals=binom.cdf(xvals,100,0.2)

plt.plot(xvals,yvals)
plt.title('Binom(100,0.2)')
plt.ylabel('F(x)')
plt.xlabel('x')
```

```python
xvals= np.arange(100)
yvals=geom.cdf(xvals,p=0.1)

plt.plot(xvals,yvals)
plt.title('Geometric(0.1)')
plt.ylabel('P(X<=x)')
plt.xlabel('x')
```

```python
#multi Variate normal

m=[0,0]
C=[[1,0.5],[0.5,1]]

#2D scatterplots of SAMPLES 

xvals=mvn.rvs(m,C,size=10000)

#otherwise matplotlig will make horizontal axis bigger
plt.xlim(-5,5)
plt.ylim(-5,5)

plt.gca().set_aspect('equal')

plt.scatter(xvals[:,0],xvals[:,1],alpha=0.2)
```
