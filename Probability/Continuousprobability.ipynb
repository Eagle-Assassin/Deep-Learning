```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform,expon,norm,laplace,gamma,beta
```

```python
x_vals=np.linspace(-2,5,1000)
y_vals=uniform.pdf(x_vals,loc=1,scale=2)
#pdf (x,loc=0,scale=1)
#What so 'loc' and scale mean?
#U (loc,loc+scale)
plt.plot(x_vals,y_vals)
plt.title('Uniform (1,3)')
plt.ylabel('f(x)')
plt.xlabel('x');
```

```python
x_vals=np.linspace(0,50,5000)
y_vals=expon.pdf(x_vals,loc=1,scale=10)
#pdf (x,loc=0,scale=1)
#What so 'loc' and scale mean?
#f(x)= (1/scale)*exp(-(x-loc)/scale)
#scale =1/lambda
plt.plot(x_vals,y_vals)
plt.title('Exp (1)')
plt.ylabel('f(x)')
plt.xlabel('x');
```

```python
x_vals=np.linspace(0,50,5000)
y_vals=norm.pdf(x_vals,loc=25,scale=10)
#pdf (x,loc=0,scale=1)
#What so 'loc' and scale mean?
#N(loc,scale)
#scale =sigma , loc = mean
plt.plot(x_vals,y_vals)
plt.title('N ^2')
plt.ylabel('f(x)')
plt.xlabel('x');
```

```python
x_vals=np.linspace(-20,50,5000)
y_vals=laplace.pdf(x_vals,loc=10,scale=5)
#pdf (x,loc=0,scale=1)
#What so 'loc' and scale mean?
#laplace(loc,scale)
#scale =a , loc = b
plt.plot(x_vals,y_vals)
plt.title('Laplace (5,10)')
plt.ylabel('f(x)')
plt.xlabel('x');
```

```python
x_vals=np.linspace(0,50,5000)
y_vals=gamma.pdf(x_vals,10,0,1)
#pdf (x,a,loc=0,scale=1)
#Note the extra parameter 'a'
#What so 'loc' and scale mean?
#Gamma (x,a,loc,scale)=Gamma ((x-loc)/scale,a)/scale
#Where Gamma(a,loc=0,scale=1)=x6(a-1)*e^(-x)/~(a)
#scale =1/beta
#a=alpha
plt.plot(x_vals,y_vals)
plt.title('Gamma(10,1)')
plt.ylabel('f(x)')
plt.xlabel('x');
```

```python
x_vals=np.linspace(0,1,100)
a=3
b=10
y_vals=beta.pdf(x_vals,a,b)
#pdf (x,a,b,loc=0,scale=1)
#Note the extra parameter 'a' and b
#What so 'loc' and scale mean?
#Lets just ignore
plt.plot(x_vals,y_vals)
plt.title('Beta(3,10)')
plt.ylabel('f(x)')
plt.xlabel('x');
```
