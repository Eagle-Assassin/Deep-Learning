```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli, binom, geom, poisson
```

```python
x_vals=np.arange(-5,5)
y_vals=bernoulli.pmf(x_vals,0.2)
plt.bar(x_vals,y_vals)
plt.title('Bernoulli(0.2)')
plt.ylabel('P(X=x)')
plt.xlabel('x')
```

```python
x_vals=np.arange(11)
y_vals= binom.pmf(x_vals,n=10,p=0.5)
plt.bar(x_vals,y_vals)
plt.title('Binomial(10,5)')
plt.ylabel('P(X=x)')
plt.xlabel('x');
```

```python
x_vals=np.arange(100)
y_vals= binom.pmf(x_vals,n=10,p=0.5)
plt.bar(x_vals,y_vals)
plt.title('Binomial(10,5)')
plt.ylabel('P(X=x)')
plt.xlabel('x');
```

```python
x_vals=np.arange(100)
y_vals= geom.pmf(x_vals,p=0.1)
#note: scipy uses the total # of trials version (vs. total failures)
#So P(X=0)=0
plt.bar(x_vals,y_vals)
plt.title('Geometric(0.1)')
plt.ylabel('P(X=x)')
plt.xlabel('x');
```

```python
x_vals=np.arange(100)
y_vals= poisson.pmf(x_vals,mu=.1)
#note: scipy uses the total # of trials version (vs. total failures)
#So P(X=0)=0
plt.bar(x_vals,y_vals)
plt.title('Poisson(0.9)')
plt.ylabel('P(X=x)')
plt.xlabel('x');
```
